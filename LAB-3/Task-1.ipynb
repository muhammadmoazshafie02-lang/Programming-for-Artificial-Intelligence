```python
# Starting Data
Model_Scores = {
    "LogisticRegression": [0.82, 0.85, 0.81, 0.86, 0.83],
    "RandomForest": [0.89, 0.91, 0.88, 0.92, 0.90],
    "NeuralNet": [0.95, 0.94, 0.93, 0.96, 0.95]
}
Best_Model = ""
Highest_Average = 0
for Model, Scores in Model_Scores.items():
    Average = sum(Scores) / len(Scores)
    if Average > Highest_Average:
        Highest_Average = Average
        Best_Model = Model
print("Best Model:", Best_Model)
print("Average Accuracy:", Highest_Average)
```
