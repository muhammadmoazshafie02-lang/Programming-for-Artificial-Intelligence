```python
Text = "AI is Transforming AI . The Future of AI is Now."
def build_vocab(Text):
    Words = Text.lower().split()
    Vocab = {}
    for Word in Words:
        if Word in Vocab:
            Vocab[Word] = Vocab[Word] + 1
        else:
            Vocab[Word] = 1
    return Vocab
print(build_vocab(Text))
```
