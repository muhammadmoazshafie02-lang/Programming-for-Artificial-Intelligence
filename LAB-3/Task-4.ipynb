```python
Graph = {
"Node_A": [{"node": "Node_B", "cost": 2}, {"node": "Node_C", "cost": 4}],
"Node_B": [{"node": "Node_A", "cost": 2}],
"Node_C": [{"node": "Node_A", "cost": 4}]
}
Graph.update({"Node_D": [{"node": "Node_A", "cost": 5}]})
Graph["Node_A"].append({"node": "Node_D", "cost": 5})
for Node in Graph:
    print(Node, ":", Graph[Node])
```
