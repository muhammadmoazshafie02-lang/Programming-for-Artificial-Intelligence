```python
print("1- Calculate Discount")
print("2- Check Membership Cashback")
print("3- Check Age Based Offer")
print("4- Exit (Bye Bye)")
Choice = int(input("Enter Your Choice: "))
match Choice:
    case 1:
            Bill=int(input("Enter Bill Amount: "))
            if Bill >= 5000:
                Discount = 20
            elif Bill >= 2000:
                Discount = 10
            elif Bill >= 1000:
                Discount = 5
            else:
                Discount = 0
            Discount_Amount = Bill * Discount / 100
            Final_Amount = Bill - Discount_Amount
            print("Discount Applied is ",str(Discount)+"%")
            print("Final Amount to Pay is ",Final_Amount)
    case 2:
            Membership_Tier = str(input("Enter Membership Tier: "))
            Amount = int(input("Enter Purchase Amount: "))
            match Membership_Tier:
                case "Gold":
                    Cashback = 15
                case "Silver":
                    Cashback = 10
                case "Bronze":
                    Cashback = 5
                case "None":
                    Cashback = 0
                case _:
                    print("Invalid")
            Cashback_Amount = Amount * Cashback/100
            print("Cashback Earned: ",Cashback_Amount)
    case 3:
            Age = int(input("Enter Age: "))
            Member = input("Are You Member or Not? ")

            if Age < 13:
                Offer = "Kids Offer: Free Toy on Purchase above $500"

            else:
                if Age < 20:
                    Offer = "Teen Offer: 10% Off on Stationery" 

                elif 20 <= Age < 60:
                    Offer = "Adult Offer: Standard Discounts Apply"

                else:
                    Offer = "Senior Citizen Offer: Extra 15% Discount"

            if Member == "Yes":
                Offer = Offer + " + Priority Billing Counter"

            print(Offer)
    case _:
        print("Exit (Bye Bye)")
```
