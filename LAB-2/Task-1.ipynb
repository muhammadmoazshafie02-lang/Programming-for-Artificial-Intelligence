```python
Bill=int(input("Enter Bill Amount: "))
if Bill >= 5000:
    Discount = 20
elif Bill >= 2000:
    Discount = 10
elif Bill >= 1000:
    Discount = 5
else:
    Discount = 0
Discount_Amount = Bill * Discount / 100
Final_Amount = Bill - Discount_Amount
print("Discount Applied is ",str(Discount)+"%")
print("Final Amount to Pay is ",Final_Amount)
```
