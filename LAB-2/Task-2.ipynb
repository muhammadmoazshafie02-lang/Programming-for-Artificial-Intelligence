```python
Membership_Tier = str(input("Enter Membership Tier: "))
Amount = int(input("Enter Purchase Amount: "))
match Membership_Tier:
    case "Gold":
        Cashback = 15
    case "Silver":
        Cashback = 10
    case "Bronze":
        Cashback = 5
    case "None":
        Cashback = 0
    case _:
        print("Invalid")
Cashback_Amount = Amount * Cashback/100
print("Cashback Earned: ",Cashback_Amount)
```
